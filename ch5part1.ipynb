````python
gvar = 100

def myfunction1():
	print ('Inside the function ', gvar)


def main():
	global gvar
	print ('Before call myfunction1 ', gvar)
	gvar += 11

	myfunction1()
	print ('After call myfunction1 ', gvar)

if __name__ == '__main__':
	main()
````

####  Explain the printed the value of "gvar"

Next gvar's new value is printed, and within the main() function, after being passed through myFunction1() its value is further increased in function main(). finally its printed again


````python
gvar = 100

def myfunction1():
    global gvar
    gvar += 1

def main():
	global gvar
	print ('Before call myfunction1 ', gvar)
	myfunction1()
	print ('After call myfunction1 ', gvar)
	gvar += 1
	print ('After increasing by 1 in main function', gvar)

if __name__ == '__main__':
	main()
````

<p> Make a function to return the generator for the Fibonacci number </p>



````python

def fibo(N):
	n1, n2 = 0, 1
	i = 0
	while ( i <= N):
		yield n1
		n3 = n1 + n2
		n1 = n2
		n2 = n3
		i += 1

gen = fibo(1)
for v in gen:
	print (v)
````

````python
def retlam( value):
	return lambda parameter : parameter * value

mylambda = retlam(10)
print (mylambda(20))

mylambda = retlam(20)
print (mylambda(20))
````

### Lambda Examples

````python
import random

lst = [ random.randint(0,100) for i in range(5)]
print (lst)

fmin = lambda x : (v for v in x if v > 50)

for v in fmin(lst):
	print (v)
````

````python
total = 0
def addValue(value):
	global total
	total += value
	return total
def subtractValue(value):
	global total
	total -= value
	return total

mylambda = lambda x, f: f(x)


print (mylambda(10, addValue))
print (mylambda(5, subtractValue))

````
